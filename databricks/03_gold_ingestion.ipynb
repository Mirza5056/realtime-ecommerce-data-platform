```python
from pyspark.sql.functions import *
```

```python
gold_df = spark.readStream.table("ecommerce.silver.silver_events")
```

```python
gold_df = (
    gold_df
    .groupBy("product_id")
    .agg(
        sum("qty").alias("units_sold"),
        sum("amount").alias("revenue"),
        countDistinct("user_id").alias("unique_customers")
    )
)
```

```python
gold_df.writeStream \
    .outputMode("append") \
    .option("checkpointLocation","/Volumes/file_upload/files/tmp/delta/gold_event_checkpoint/") \
    .trigger(availableNow=True) \
    .toTable("ecommerce.gold.gold_metrics")
```
